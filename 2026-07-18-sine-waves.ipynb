# Sine waves, interactively

A demo post. Click the power button (bottom of the page or per cell) to start an in-browser Python kernel — then edit the code and re-run it. Nothing leaves your machine.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 4 * np.pi, 400)
for k in (1, 2, 3):
    plt.plot(x, np.sin(k * x) / k, label=f"sin({k}x)/{k}")
plt.legend()
plt.title("Try changing k — this runs in your browser")
plt.show()
```
