# Sine waves, interactively

How the partial sums of $\sin(kx)/k$ converge as more terms are added. Drag the slider to change how many terms are summed — the chart updates instantly in your browser. To change the code itself, open the editable notebook linked at the bottom.

```python
import numpy as np
import plotly.graph_objects as go

x = np.linspace(0, 4 * np.pi, 400)

fig = go.Figure()
for k in range(1, 9):
    y = sum(np.sin(i * x) / i for i in range(1, k + 1))
    fig.add_trace(go.Scatter(x=x, y=y, visible=(k == 3), name=f"k={k}"))

steps = [
    dict(method="update",
         args=[{"visible": [j == k - 1 for j in range(8)]},
               {"title": f"Partial sum of the first {k} terms"}],
         label=str(k))
    for k in range(1, 9)
]
fig.update_layout(
    sliders=[dict(active=2, currentvalue={"prefix": "terms: "}, steps=steps)],
    title="Partial sum of the first 3 terms",
    margin=dict(t=60),
)
fig.show()
```
